```python
# 1. Create a Pandas Series containing the marks
import pandas as pd
marks = [78, 85, 92, 67, 88]
series = pd.Series(marks)

print("Complete Series",series)
print("Values",series.values)
print("Index",series.index)
print("Data type",series.dtype)
print("Mean",series.mean())
print("Maximum",series.max())
print("Minimum",series.min())
```

```python
# 2. Create a Series containing:
marks_series = pd.Series([85, 78, 92, 74, 88], index=['Asha', 'Ravi', 'Meena', 'Kiran', 'Arun'])

print("Marks of Meena:", marks_series['Meena'])
```

```python
# 3. Create the following DataFrame:

data = {
    'Name': ['Asha', 'Ravi', 'Meena', 'Kiran', 'Arun'],
    'Age': [22, 23, 21, 24, 22],
    'Department': ['CSE', 'ECE', 'CSE', 'ME', 'ECE'],
    'Marks': [85, 78, 92, 74, 88]
}

df = pd.DataFrame(data)
print("Complete DataFrame",df)
print("First 3 Rows",df.head(3))
print("Last 2 Rows",df.tail(2))
print("Shape",df.shape)
print("Column Names",df.columns)
print("Data Types",df.dtypes)
```

```python
# 4. Using the previous DataFrame, display:
print("Info:",df.info())
print("Describe():",df.describe())
print("Number of rows:", df.shape[0]) 
print("Number of columns:", df.shape[1])
```

```python
# 5. Using df, display:
print("Name column:\n", df['Name'])
print("Marks column:\n", df['Marks'])
print("Name and Marks together:\n", df[['Name', 'Marks']])
```

```python
# 6. Display:
print("First row using iloc:\n", df.iloc[0])
print("\nThird row using iloc:\n", df.iloc[2])
print("\nRows 1 to 3 (index 1 to 3 exclusive):\n", df.iloc[1:4])
print("\nName and Marks columns using loc:\n", df.loc[:, ['Name', 'Marks']])
```

```python
# 7. Find:
print("Students scoring above 80:\n", df[df['Marks'] > 80])
print("\nStudents scoring between 75 and 90:\n", df[df['Marks'].between(75, 90)])
print("\nStudents belonging to CSE:\n", df[df['Department'] == 'CSE'])
print("\nCSE students scoring above 80:\n", df[(df['Department'] == 'CSE') & (df['Marks'] > 80)])
```

```python
# 8. Display students belonging to either CSE or ECE.
print(df[df['Department'].isin(['CSE', 'ECE'])])
```

```python
# 9. Add a new column called Result. Students scoring 75 or above should receive "Pass", otherwise "Fail".
df['Result'] = ['Pass' if m >= 75 else 'Fail' for m in df['Marks']]
print(df)
```

```python
# 10. Assign grades according to:
def assign_grade(marks):
    if marks >= 90:
        return 'A'
    elif marks >= 80:
        return 'B'
    elif marks >= 70:
        return 'C'
    else:
        return 'D'

df['Grade'] = df['Marks'].apply(assign_grade)
print(df)
```

```python
# 11. Sort students:
print("By marks in ascending order:\n", df.sort_values(by='Marks', ascending=True))
print("\nBy marks in descending order:\n", df.sort_values(by='Marks', ascending=False))
print("\nBy Department and then Marks:\n", df.sort_values(by=['Department', 'Marks'], ascending=[True, True]))
```
