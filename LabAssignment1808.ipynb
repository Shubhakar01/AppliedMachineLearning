```python
# 3. Create the following DataFrame:

data = {
    'Name': ['Asha', 'Ravi', 'Meena', 'Kiran', 'Arun'],
    'Age': [22, 23, 21, 24, 22],
    'Department': ['CSE', 'ECE', 'CSE', 'ME', 'ECE'],
    'Marks': [85, 78, 92, 74, 88]
}

df = pd.DataFrame(data)
print("Complete DataFrame",df)
print("First 3 Rows",df.head(3))
print("Last 2 Rows",df.tail(2))
print("Shape",df.shape)
print("Column Names",df.columns)
print("Data Types",df.dtypes)
```

```python
# 4. Using the previous DataFrame, display:
print("Info:",df.info())
print("Describe():",df.describe())
print("Number of rows:", df.shape[0]) 
print("Number of columns:", df.shape[1])
```

```python
# 5. Using df, display:
print("Name column:\n", df['Name'])
print("Marks column:\n", df['Marks'])
print("Name and Marks together:\n", df[['Name', 'Marks']])
```

```python
# 6. Display:
print("First row using iloc:\n", df.iloc[0])
print("\nThird row using iloc:\n", df.iloc[2])
print("\nRows 1 to 3 (index 1 to 3 exclusive):\n", df.iloc[1:4])
print("\nName and Marks columns using loc:\n", df.loc[:, ['Name', 'Marks']])
```

```python
# 7. Find:
print("Students scoring above 80:\n", df[df['Marks'] > 80])
print("\nStudents scoring between 75 and 90:\n", df[df['Marks'].between(75, 90)])
print("\nStudents belonging to CSE:\n", df[df['Department'] == 'CSE'])
print("\nCSE students scoring above 80:\n", df[(df['Department'] == 'CSE') & (df['Marks'] > 80)])
```

```python
# 8. Display students belonging to either CSE or ECE.
print(df[df['Department'].isin(['CSE', 'ECE'])])
```

```python
# 9. Add a new column called Result. Students scoring 75 or above should receive "Pass", otherwise "Fail".
df['Result'] = ['Pass' if m >= 75 else 'Fail' for m in df['Marks']]
print(df)
```

```python
# 10. Assign grades according to:
def assign_grade(marks):
    if marks >= 90:
        return 'A'
    elif marks >= 80:
        return 'B'
    elif marks >= 70:
        return 'C'
    else:
        return 'D'

df['Grade'] = df['Marks'].apply(assign_grade)
print(df)
```

```python
# 11. Sort students:
print("By marks in ascending order:\n", df.sort_values(by='Marks', ascending=True))
print("\nBy marks in descending order:\n", df.sort_values(by='Marks', ascending=False))
print("\nBy Department and then Marks:\n", df.sort_values(by=['Department', 'Marks'], ascending=[True, True]))
```

```python
# 12. Consider:
import numpy as np
import pandas as pd

data = {
    "Name": ["Asha", "Ravi", "Meena", "Kiran", "Arun"],
    "Age": [22, 23, np.nan, 24, 22],
    "Marks": [85, np.nan, 92, 74, 88],
}
students = pd.DataFrame(data)

# Find missing values (boolean mask)
print(students.isnull())

# Number of missing values in each column
print(students.isnull().sum())

# Rows containing missing values
print(students[students.isnull().any(axis=1)])
```

```python
# 13. Replace:
students["Age"] = students["Age"].fillna(students["Age"].mean())
students["Marks"] = students["Marks"].fillna(students["Marks"].mean())
print(students)
```

```python
#14. Remove rows containing missing values.
clean_students = students.dropna()
print(clean_students)
```

```python
#17. Consider:
data = {
    "Name": ["Asha", "Ravi", "Meena", "Kiran", "Arun", "Divya"],
    "Department": ["CSE", "ECE", "CSE", "ME", "ECE", "CSE"],
    "Marks": [85, 78, 92, 74, 88, 80],
}
df = pd.DataFrame(data)

# Average, Maximum, and Count by department
print("Average Marks:\n", df.groupby("Department")["Marks"].mean())
print("Maximum Marks:\n", df.groupby("Department")["Marks"].max())
print("Student Count:\n", df.groupby("Department")["Name"].count())

# Comprehensive summary using agg()
print(
    "\nDepartment Statistics:\n",
    df.groupby("Department")["Marks"].agg(["mean", "min", "max", "count"]),
)
```

```python
#19. Use the following dataset:
data = {
    "Name": ["Asha", "Ravi", "Meena", "Kiran", "Arun", "Divya", "John", "Priya"],
    "Department": ["CSE", "ECE", "CSE", "ME", "ECE", "CSE", "ME", "ECE"],
    "Attendance": [92, 75, 96, 68, 82, 88, 72, 95],
    "Internal": [85, 72, 94, 65, 80, 86, 70, 91],
    "FinalMarks": [88, 74, 96, 62, 83, 89, 68, 93],
}
df = pd.DataFrame(data)

print("1. First five records:\n", df.head())
print("2. Dimensions:", df.shape)
print("3. Average final mark:", df["FinalMarks"].mean())
print("4. Highest-scoring student:\n", df.loc[df["FinalMarks"].idxmax()])
print("5. Lowest-scoring student:\n", df.loc[df["FinalMarks"].idxmin()])
print("6. Students scoring > 80:\n", df[df["FinalMarks"] > 80])
print("7. Average marks by department:\n", df.groupby("Department")["FinalMarks"].mean())
print("8. Attendance > 80%:\n", df[df["Attendance"] > 80])
print("9. Sorted by FinalMarks:\n", df.sort_values(by="FinalMarks", ascending=False))

df["Result"] = df["FinalMarks"].apply(lambda x: "Pass" if x >= 70 else "Fail")


def categorize(marks):
    if marks >= 90:
        return "Excellent"
    elif marks >= 80:
        return "Very Good"
    elif marks >= 70:
        return "Good"
    else:
        return "Needs Improvement"


df["Category"] = df["FinalMarks"].apply(categorize)

dept_avg = df.groupby("Department")["FinalMarks"].mean()
print("Department with highest avg FinalMarks:", dept_avg.idxmax())
```

```python
# 20. Create:
employees = pd.DataFrame({
    "Employee": ["E01", "E02", "E03", "E04", "E05", "E06", "E07", "E08"],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "Finance", "IT", "Finance"],
    "Experience": [5, 3, 7, 4, 6, 8, 2, 5],
    "Salary": [60000, 45000, 75000, 65000, 55000, 85000, 48000, 70000],
})

print("Average Salary:", employees["Salary"].mean())
print("Maximum Salary:", employees["Salary"].max())
print("Minimum Salary:", employees["Salary"].min())
print("Highest-earning Employee:\n", employees.loc[employees["Salary"].idxmax()])
print("Employees earning > ₹60,000:\n", employees[employees["Salary"] > 60000])
print("Average salary by department:\n", employees.groupby("Department")["Salary"].mean())
print("Average experience by department:\n", employees.groupby("Department")["Experience"].mean())
print("Number of employees in each department:\n", employees.groupby("Department")["Employee"].count())
print("Sorted by salary:\n", employees.sort_values(by="Salary", ascending=False))
print("IT employees with > 3 yrs experience:\n", employees[(employees["Department"] == "IT") & (employees["Experience"] > 3)])

# Add SalaryLevel column
def salary_level(sal):
    if sal >= 75000:
        return "High"
    elif sal >= 50000:
        return "Medium"
    else:
        return "Low"

employees["SalaryLevel"] = employees["Salary"].apply(salary_level)
print(employees)
```
