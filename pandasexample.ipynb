```python
X = dataset.iloc[:,:-1].values
y = dataset.iloc[:,3].values
```

```python
# Whatever is NAN values it is replaced or filled with mean value
imputer = SimpleImputer(missing_values=np.nan, strategy='mean')
imputer = imputer.fit(X[:,1:3])
X[:,1:3] = imputer.transform(X[:,1:3])
print(X)
```

```python
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
labelencoder_X = LabelEncoder()
X[:,0]= labelencoder_X.fit_transform(X[:, 0 ])
print(X)
```

```python
# Encoding the independent Variable
from sklearn.compose import ColumnTransformer
ct = ColumnTransformer([("Country",OneHotEncoder(),[0])],remainder="passthrough")
x = ct.fit_transform(X)

x.astype(int)
```

```python
# Encoding the dependent Variable
labelencoder_y = LabelEncoder()
y = labelencoder_y.fit_transform(y)
y
```

```python
# Split data into training set and Test set
# Randomly select the datapoints from the dataset amd not in order
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=45) #Random state not given then changes each time
```
