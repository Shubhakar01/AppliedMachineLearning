```python
test = np.random.randint(0,10,(4,3))
test
```

```python
# Iterate by the row
for row in test:
    print(row)
```

```python
for i, row in enumerate(test):
    print('row',i,'is',row)
```

```python
#Use zip to iterate over multiple iterables
test2 =test**2
test2
```

```python
for i,j in zip(test,test2):
    print(i,'+',j,'=',i+j)
```

```python
marks = np.array([
    [80,75,98],
    [65,70,72],
    [88,92,85],
    [55,68,58]
])

student_total = np.sum(marks, axis = 1)
print("Total of each student:", student_total)

student_average = np.mean(marks, student_total)
print("Average of each student: ",student_average)

print("Average of each subject")

highest_student = np.argmax(student_total)
print("")

highest_subject = np.max(marks,axis=0)
print("Highest marks in each subject")
```
