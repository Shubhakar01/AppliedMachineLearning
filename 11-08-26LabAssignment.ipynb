```python
# 8. Given,
salary = np.array([25000, 32000, 28000, 45000, 50000, 38000, 42000, 30000])

#Total Salary
print(salary.sum())
#Average Salary
print(salary.mean())
#Highest Salary
print(salary.max())
#Lowest Salary
print(salary.min())
#Median Salary
print(np.median(salary))
#Standard Deviation
print(salary.std())
```

```python
# 9. Using NumPy indexing, find
marks = np.array([45,78,89,32,67,90,55,76,41,84]) 

# Marks greater than 75
print(marks[marks>75])

#Marks between 50 and 80
print(marks[(marks>=50) & (marks<=80)])

# No. of students who scored below 50
print(len(marks[marks<50]))

# Percentage of students who scored 60 or above
print(len(marks[marks>=60])*100 / len(marks))
```

```python
# 10. Following array contains marks of four students in three subjects:
marks = np.array([
    [80,75,90],
    [65,70,72],
    [88,92,85],
    [55,60,58]
])

# total marks of each student
print("Total marks of Student 1:",marks[0].sum())
print("Total marks of Student 2:",marks[1].sum())
print("Total marks of Student 3:",marks[2].sum())
print("Total marks of Student 4:",marks[3].sum())

#Average marks of each student
print("Average marks of Student 1:",marks[0].mean())
print("Average marks of Student 2:",marks[1].mean())
print("Average marks of Student 3:",marks[2].mean())
print("Average marks of Student 4:",marks[3].mean())

#Average marks in each subject
print("Average marks in Subject 1:",marks[:,0].mean())
print("Average marks in Subject 2:",marks[:,1].mean())
print("Average marks in Subject 3:",marks[:,2].mean())

#Student with the highest total

#Highest marks in each subject
print("Highest marks in Subject 1:",marks[:,0].max())
print("Highest marks in Subject 2:",marks[:,1].max())
print("Highest marks in Subject 3:",marks[:,2].max())
```

```python
# 12. Daily temperature for 2 weeks are:

temperature = np.array([
    31,32,33,35,36,34,32,
    30,29,31,33,35,37,36
])
#Mean temperature
print(temperature.mean())
# Maximum and Minimum Temperature
print(temperature.max())
print(temperature.min())
# No. of days above the average temperature
print(len(temperature[temperature>temperature.mean()]))
# Temperature greater than 34
print(temperature[temperature>34])
# Differnce between max and min
print(temperature.max() - temperature.min())
#Standard Deviation
print(temperature.std())
```
