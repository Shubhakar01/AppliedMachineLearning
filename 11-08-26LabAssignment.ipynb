```python
# 13. Monthly sales of a company are:
sales = np.array([
    12000, 15000, 17000, 14000,
    18000, 21000, 19000, 22000,
    25000, 23000, 26000, 28000
])

#Total annual sales
print(np.sum(sales))
#Average monthly sales
print(np.mean(sales))
#Month with the highest sales
print(np.argmax(sales) + 1)
#Month with the lowest sales
print(np.argmin(sales) + 1)
#Number of months with sales above ₹20,000
print(np.sum(sales > 20000))
#Percentage increase from January to December
percent_increase = ((sales[-1] - sales[0]) / sales[0]) * 100
print(f'{percent_increase:.2f}%')
```

```python
#14. Consider marks of five students in four subjects:

import numpy as np

# Sample Data: 5 Students x 4 Subjects
marks = np.array([
    [78, 85, 90, 88],
    [65, 70, 72, 68],
    [92, 88, 95, 91],
    [55, 60, 58, 62],
    [80, 75, 82, 79]
])

# Total marks of every student
print(np.sum(marks, axis=1))
# Average of every student
student_averages = np.mean(marks, axis=1)
print(student_averages)
# Average mark for every subject
print(np.mean(marks, axis=0))
# Highest-scoring student (0-indexed position)
print(np.argmax(student_averages))
# Lowest-scoring student (0-indexed position)
print(np.argmin(student_averages))
# Students whose average is greater than 80 (indices/averages)
print(student_averages[student_averages > 80])
# Overall class average
print(np.mean(marks))
# Assign categories using np.select()
conditions = [
    student_averages >= 85,
    (student_averages >= 70) & (student_averages < 85),
    (student_averages >= 50) & (student_averages < 70),
    student_averages < 50
]
categories = ["Excellent", "Good", "Average", "Needs Improvement"]

performance_categories = np.select(conditions, categories,default="N/A")
print(performance_categories)
```

```python
# 21. Take your own example to normalize an array.  
data = np.array([12, 23, 34, 45, 56], dtype=float)
data_normalized = (data - np.min(data)) / (np.max(data) - np.min(data))
print(data_normalized)
```

```python
# 23. Create a NumPy array containing 100 random exam marks and 
# determine how many students fall into the ranges 0–39, 40–59, 60–79, and 80–100.

exam_marks = np.random.randint(0, 101, size=100)

bins = [0, 40, 60, 80, 101]
labels = ["0–39", "40–59", "60–79", "80–100"]
counts, _ = np.histogram(exam_marks, bins=bins)

for label, count in zip(labels, counts):
    print(f"Range {label}: {count} students")
```

```python
# 24. Compare the time required to multiply 1,000,000 numbers 
# using a Python list and a NumPy array.

import time

size = 1_000_000

# Setup Data
py_list1 = list(range(size))
py_list2 = list(range(size))

np_arr1 = np.arange(size, dtype=np.int64)
np_arr2 = np.arange(size, dtype=np.int64)

# Native Python List Benchmark
start_time = time.time()
py_result = [x * y for x, y in zip(py_list1, py_list2)]
py_time = time.time() - start_time

# NumPy Array Benchmark
start_time = time.time()
np_result = np_arr1 * np_arr2
np_time = time.time() - start_time

print(f"Python List Time: {py_time:.6f} seconds")
print(f"NumPy Array Time: {np_time:.6f} seconds")
print(f"NumPy is ~{py_time / np_time:.2f}x faster.")
```
